```python
hyperopt asal m phly kuch random hyperparameter combinations pr model ko train kr test score check krta h
and then un score results s learn krta rehta h, yani y dekhta h k agly kon s combinations uthay jay k un ka score ziada ho ga

from hyperopt import fmin, tpe, hp, Trials :::::::

fmin: is k ander kuch hyperparameters jaty hn jis s sab s phly objective function jata h yani vo function jis k ander
        model train hona hota h
tpe:  (Tree of Parzen Estimators) yhi function h asal m jo k previous combination results s learn krta h and btata h k 
        agly in hyperparameter combinations pr train kro q k in s ziada score any ki probability h yani y probablity ki base
        pr agly combinations ko btata h

Search space :  is k ander hm hyperparametrs k name dety hn jo k hm n different values pr test krny hoty hn 
hp:      they are search space functions yani 
            hp.uniform (for continious uniform distribution), 
            hp.quniform() for discrete uniform distribution (e.g., integers)
            hp.choice: For categorical variables.

Trials:   trails function k ander saary combination k results hoty hn to fmin us trials vay m s best result utha kr deta h
```

```python
from hyperopt import fmin, tpe, hp, Trials
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

# Load a sample dataset
data = load_iris()
X = data.data
y = data.target

# Define the objective function
def objective(params):
    # Convert n_estimators to an integer
    params['n_estimators'] = int(params['n_estimators'])
    model = RandomForestClassifier(**params)
    score = cross_val_score(model, X, y, cv=5).mean()
    return -score  # We want to maximize accuracy, so minimize the negative accuracy

# Define the search space
space = {
    'n_estimators': hp.quniform('n_estimators', 10, 200, 5), # yani hr bar 5 ka gap lo, phly 10, phir 15 phir 20 ...
    # 'max_depth': hp.quniform('max_depth', 2, 20, 1),
    'criterion': hp.choice('criterion', ['gini', 'entropy'])
}

# Run the optimization
trials = Trials()
best = fmin(fn=objective, space=space, algo=tpe.suggest, max_evals=100, trials=trials)

print("Best Hyperparameters:", best)
```
